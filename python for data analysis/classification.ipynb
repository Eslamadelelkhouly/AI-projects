```python
y = Data['carrier']
x = Data[['arr_flights', 'arr_del15', 'carrier_ct', 'weather_ct', 'nas_ct',
       'security_ct', 'late_aircraft_ct', 'arr_cancelled', 'arr_diverted',
       'arr_delay', 'carrier_delay', 'weather_delay', 'nas_delay',
       'security_delay', 'late_aircraft_delay']]
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , Y_train , Y_test = train_test_split(x,y,test_size=0.20 , random_state=44 , shuffle=True)
print("X_train shape = ",X_train.shape)
print("X_test shape = ",X_test.shape)
print("Y_train shape = ",Y_train.shape)
print("Y_test shape = ",Y_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
randomforestclassifiermodel = RandomForestClassifier(criterion='gini' , n_estimators=100 , max_depth=2 , random_state=33)
```

```python
randomforestclassifiermodel.fit(X_train ,Y_train)
```

```python
y_pred = randomforestclassifiermodel.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import classification_report
```

```python
ClassificationReports = classification_report(Y_test,y_pred)
print('Classification Report is : \n', ClassificationReports)
```
